```python
base_path = "/Volumes/workspace/default/fmcg_data/"
bronze_path = base_path + "bronze/"
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
# Reading all csv files and writing to delta
```

```python
customers_a = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .option("delimiter", ",")
    .load("/Volumes/workspace/default/fmcg_data/company_a/customers_A.csv")
)
```

```python
customers_a.write \
    .format("delta") \
    .mode("overwrite") \
    .save(bronze_path + "customers_a")
```
