**Q1: What are the key limitations of traditional MapReduce that make Spark a preferred
choice for modern big data processing?**

Ans 1: Traditional MapReduce stores intermediate results on disk after every processing step, which makes it slower, especially for large datasets. It is not suitable for iterative tasks like machine learning because the same data has to be read from disk repeatedly.
Spark overcomes these problems by processing data in memory, resulting in much faster execution and easier development.

**Q2: Explain how Spark uses In-Memory Computing to speed up iterative machine learning
algorithms compared to disk-based systems.**

Ans 2:

*   Stores data in RAM instead of disk.
*   Reduces disk read/write operations.
*   Reuses data across multiple iterations.

**Q9: When cleaning a dataset, why is it often better to handle null values before
performing mathematical aggregations like sum() or avg()?**

Ans 9:  Handling null values before aggregation improves the accuracy of the results. If null values are ignored or left untreated, calculations like average or sum may produce incorrect or misleading results.

**Q11: Explain the "Shuffle" process that occurs during a grouping operation. Why is it
considered a wide transformation?**

Ans 11: Shuffle is the process where Spark redistributes data across different partitions so that records with the same key are placed together. This usually happens during operations like groupBy() or joins.

It is called a wide transformation because data moves between different partitions and executors across the cluster

**Q13: How do you use the .agg() function to calculate multiple statistics at once, such as
the min, max, and mean of the price column?**

```python
df3 = spark.read.csv(
    "store_sales_small.csv",
    header=True,
    inferSchema=True
)
```

```python
df3.agg(
    min("price").alias("min_price"),
    max("price").alias("max_price"),
    mean("price").alias("avg_price")
).show()
```

**Q14: In the context of cleaning a dataset, what is the risk of using inferSchema=true
when your source data contains messy or inconsistent date formats?**

Ans 14:
* May infer incorrect data types.
* Inconsistent date formats can be treated as strings.
* Can cause parsing and conversion errors.
* May affect filtering and sorting.
* Defining the schema manually is more reliable for messy data.

**Q15: Write a final processing pipeline that:**
**1. Filters out duplicates.
2. Fills null prices with 0.
3. Groups by store_id to calculate total revenue.**

```python

result = (
    df3.dropDuplicates()
      .fillna({"price": 0})
      .groupBy("store_id")
      .agg(sum("price").alias("total_revenue"))  #total revenue
)

result.show()
```
